## 2.2 Duplicate Detection

```python
duplicate_rows = df.duplicated().sum()

print("Number of Duplicate Rows:", duplicate_rows)
```

```python
duplicate_order_ids = df['Order ID'].duplicated().sum()

print("Duplicate Order IDs:", duplicate_order_ids)
```

### Interpretation

The duplicate row analysis showed that the dataset contains no fully duplicated rows. However, there are 4985 duplicate Order IDs. This is expected in a retail transactional dataset because a single customer order may contain multiple products. ## 2.3 Data Type Conversion

```python
df['Order Date'] = pd.to_datetime(df['Order Date'])

df['Ship Date'] = pd.to_datetime(df['Ship Date'])

print("Date columns converted successfully!")
```

```python
df['Shipping_Days'] = (df['Ship Date'] - df['Order Date']).dt.days

df[['Order Date', 'Ship Date', 'Shipping_Days']].head()
```

```python
df['Profit_Margin'] = (df['Profit'] / df['Sales']) * 100

df[['Sales', 'Profit', 'Profit_Margin']].head()
```

```python
df['Year'] = df['Order Date'].dt.year

df['Month'] = df['Order Date'].dt.month

df[['Order Date', 'Year', 'Month']].head()
```

```python
Q1_sales = df['Sales'].quantile(0.25)

Q3_sales = df['Sales'].quantile(0.75)

IQR_sales = Q3_sales - Q1_sales

print("Q1:", Q1_sales)
print("Q3:", Q3_sales)
print("IQR:", IQR_sales)
```

```python
lower_bound_sales = Q1_sales - (1.5 * IQR_sales)

upper_bound_sales = Q3_sales + (1.5 * IQR_sales)

print("Lower Bound:", lower_bound_sales)
print("Upper Bound:", upper_bound_sales)
```

```python
sales_outliers = df[
    (df['Sales'] < lower_bound_sales) |
    (df['Sales'] > upper_bound_sales)
]

print("Number of Sales Outliers:", sales_outliers.shape[0])
```

```python
plt.figure(figsize=(10,5))

sns.boxplot(x=df['Sales'])

plt.title("Sales Distribution Boxplot")

plt.xlabel("Sales")

plt.show()
```

```python
Q1_profit = df['Profit'].quantile(0.25)

Q3_profit = df['Profit'].quantile(0.75)

IQR_profit = Q3_profit - Q1_profit

print("Q1:", Q1_profit)
print("Q3:", Q3_profit)
print("IQR:", IQR_profit)
```

```python
lower_bound_profit = Q1_profit - (1.5 * IQR_profit)

upper_bound_profit = Q3_profit + (1.5 * IQR_profit)

print("Lower Bound:", lower_bound_profit)
print("Upper Bound:", upper_bound_profit)
```

```python
profit_outliers = df[
    (df['Profit'] < lower_bound_profit) |
    (df['Profit'] > upper_bound_profit)
]

print("Number of Profit Outliers:", profit_outliers.shape[0])
```

```python
plt.figure(figsize=(10,5))

sns.boxplot(x=df['Profit'])

plt.title("Profit Distribution Boxplot")

plt.xlabel("Profit")

plt.show()
```

### Outlier Treatment Decision

The outlier analysis identified several extreme values in both the Sales and Profit columns. However, these outliers were retained rather than removed or capped. In a retail business context, unusually high sales values may represent legitimate bulk purchases, corporate transactions, or premium product sales rather than data errors.

Similarly, extreme profit and loss values may provide important business insights regarding highly profitable products or problematic transactions generating substantial losses. Removing these records could distort the true business performance and reduce the accuracy of profitability analysis.

Therefore, all outliers were retained to preserve the integrity and realism of the dataset for further analysis and dashboard reporting.

## 3.2 Grouped Analysis

```python
category_analysis = df.groupby('Category').agg({
    'Sales': 'sum',
    'Profit': 'sum',
    'Discount': 'mean'
})

print(category_analysis)
```

### Category Analysis Interpretation

The grouped analysis shows that Technology generated the highest total sales and profit among all product categories, indicating strong overall performance and profitability. Office Supplies also contributed positively to profit, although with lower sales compared to Technology.

Furniture generated relatively high sales but lower profitability compared to the other categories. This suggests that despite strong sales performance, factors such as high discounts or operational costs may be reducing profit margins within the Furniture category.

The average discount values indicate that discounting practices vary across categories, which may influence profitability outcomes.

```python
plt.figure(figsize=(8,5))
sns.barplot(x='Category', y='Sales', data=df)

plt.title("Sales by Category")

plt.show()
```

Sales by Category Visualization Interpretation

The visualization shows that Technology achieved the highest sales among all categories, followed by Furniture and Office Supplies. This supports the earlier grouped analysis findings and highlights Technology as the strongest-performing category in terms of revenue generation.

```python
region_segment_analysis = df.groupby(['Region', 'Segment']).agg({
    'Sales': 'sum',
    'Profit': 'sum'
})

print(region_segment_analysis)
```

### Regional and Segment Analysis Interpretation

The regional and segment analysis highlights differences in sales and profitability across geographic regions and customer groups. The West and East regions generally demonstrate stronger sales and profit performance compared to the Central and South regions.

Consumer customers contribute significantly to total sales across most regions, while Corporate and Home Office segments show varying profitability levels. Some regional and segment combinations generate lower profits despite reasonable sales performance, suggesting potential operational inefficiencies or excessive discounting in certain markets.

```python
region_profit = df.groupby('Region')['Profit'].sum()

region_profit.plot(kind='bar')

plt.title("Profit by Region")

plt.show()
```

Profit by Region Chart Interpretation

The bar chart shows differences in profitability across regions. The West region generated the highest overall profit, while some regions achieved comparatively lower profits. This indicates that regional performance varies and suggests opportunities for targeted business strategies to improve profitability in weaker-performing regions.

```python
sub_category_profit_margin = df.groupby('Sub-Category')['Profit_Margin'].mean().sort_values()

print(sub_category_profit_margin)
```

### Sub-Category Profit Margin Interpretation

The sub-category analysis reveals significant differences in profitability across products. Certain sub-categories, such as Tables and Bookcases, show very low or negative average profit margins, indicating that these products may be generating losses for the company.

In contrast, some sub-categories achieve strong positive profit margins, suggesting better pricing strategies and cost efficiency. The findings indicate that management should closely review underperforming sub-categories to determine whether pricing, discount strategies, or operational costs are contributing to poor profitability.

```python
monthly_sales_trend = df.groupby(['Year', 'Month'])['Sales'].sum()

print(monthly_sales_trend)
```

## 3.3 Correlation Analysis

```python
correlation_matrix = df[['Sales', 'Quantity', 'Discount', 'Profit']].corr()

print(correlation_matrix)
```

```python
plt.figure(figsize=(8,6))

sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm')

plt.title("Correlation Matrix Heatmap")

plt.show()
```

### Correlation Analysis Interpretation

The correlation analysis shows different relationships between the numerical variables in the dataset. The strongest positive correlation exists between Sales and Profit, indicating that higher sales transactions generally contribute to higher profits.

A negative correlation exists between Discount and Profit, suggesting that higher discount levels tend to reduce profitability. This indicates that aggressive discount strategies may be eroding profit margins and contributing to losses in certain transactions.

The relationship between Quantity and Profit appears relatively weak, suggesting that simply selling larger quantities does not necessarily guarantee higher profitability.

## 3.4 Profitability Investigation

```python
loss_transactions = df[df['Profit'] < 0]

loss_transactions.head()
```

```python
total_loss_amount = loss_transactions['Profit'].sum()

number_of_losses = loss_transactions.shape[0]

print("Total Loss Amount:", total_loss_amount)

print("Number of Loss-Making Transactions:", number_of_losses)
```

```python
sub_category_losses = loss_transactions.groupby('Sub-Category')['Profit'].sum().sort_values()

print(sub_category_losses)
```

### Profitability Investigation Interpretation

The profitability investigation identified a significant number of transactions operating at a loss. The total loss amount highlights the negative financial impact of unprofitable sales transactions on the business.

The sub-category analysis revealed that certain products, such as Tables and Bookcases, contribute substantially to overall losses. These findings suggest that current pricing, discount strategies, or operational costs within these product lines may require review.

The analysis also supports the earlier correlation findings that excessive discounting may be negatively affecting profitability across several product categories.
